```python

import duckdb
import pandas as pd
from pathlib import Path

PROJECT_ROOT = Path(
    "/content/gdrive2/MyDrive/NFL-BigDataBowl-2027"
)

RAW_DIR = PROJECT_ROOT / "data" / "raw"
FEATURE_DIR = PROJECT_ROOT / "data" / "features"
REPORT_DIR = PROJECT_ROOT / "reports" / "data_audit"

combine_file = next(RAW_DIR.rglob("combine_tracking.csv"))
results_file = next(RAW_DIR.rglob("combine_results.csv"))

con = duckdb.connect()

# Register source data
con.execute(f"""
CREATE OR REPLACE VIEW combine_tracking AS
SELECT *
FROM read_csv_auto('{combine_file.as_posix()}')
""")

con.execute(f"""
CREATE OR REPLACE VIEW combine_results AS
SELECT *
FROM read_csv_auto('{results_file.as_posix()}')
""")

# 1. Schema
print("=" * 65)
print("COMBINE TRACKING SCHEMA")
print("=" * 65)

print(con.execute(
    "DESCRIBE combine_tracking"
).df().to_string(index=False))

# 2. Entity types
print("\nENTITY TYPE DISTRIBUTION")

print(con.execute("""
SELECT entity_type, COUNT(*) AS observations
FROM combine_tracking
GROUP BY entity_type
ORDER BY observations DESC
""").df().to_string(index=False))

# 3. Wide receiver drill inventory
print("\nWIDE RECEIVER DRILL INVENTORY")

drill_inventory = con.execute("""
SELECT
    drill_type,
    drill_name,
    COUNT(*) AS observations,
    COUNT(DISTINCT nfl_id) AS unique_players,
    COUNT(DISTINCT attempt) AS distinct_attempt_values,
    MIN(time) AS first_timestamp,
    MAX(time) AS last_timestamp
FROM combine_tracking
WHERE entity_type = 'PLAYER'
  AND nfl_id IN (
      SELECT nfl_id
      FROM combine_results
      WHERE combine_position = 'WR'
  )
GROUP BY drill_type, drill_name
ORDER BY unique_players DESC, observations DESC
""").df()

print(drill_inventory.to_string(index=False))

# Save audit
REPORT_DIR.mkdir(parents=True, exist_ok=True)

drill_inventory.to_csv(
    REPORT_DIR / "combine_wr_drill_audit.csv",
    index=False
)

# 4. Inspect actual measurements
print("\nSAMPLE TRACKING OBSERVATIONS")

print(con.execute("""
SELECT *
FROM combine_tracking
WHERE entity_type = 'PLAYER'
ORDER BY nfl_id, drill_type, time
LIMIT 15
""").df().to_string(index=False))

# 5. Combine results schema
print("\nCOMBINE RESULTS")

print(con.execute("""
SELECT *
FROM combine_results
WHERE combine_position = 'WR'
LIMIT 10
""").df().to_string(index=False))

print("\nAUDIT SAVED SUCCESSFULLY")
```

```python

attempt_quality = con.execute("""
WITH wr_players AS (
    SELECT DISTINCT nfl_id
    FROM combine_results
    WHERE combine_position = 'WR'
),

ordered AS (
    SELECT
        draft_year,
        nfl_id,
        drill_type,
        drill_name,
        attempt,
        event_id,
        time,
        x, y, s, a, dis, dir,

        LAG(time) OVER (
            PARTITION BY
                draft_year, nfl_id, drill_type,
                drill_name, attempt, event_id
            ORDER BY time
        ) AS previous_time

    FROM combine_tracking
    WHERE entity_type = 'PLAYER'
      AND nfl_id IN (SELECT nfl_id FROM wr_players)
),

gaps AS (
    SELECT *,
        EPOCH(time - previous_time) AS time_gap_seconds
    FROM ordered
)

SELECT
    draft_year,
    nfl_id,
    drill_type,
    drill_name,
    attempt,
    event_id,

    COUNT(*) AS observations,

    MIN(time) AS start_time,
    MAX(time) AS end_time,

    COUNT(*) FILTER (WHERE x IS NULL OR y IS NULL)
        AS missing_coordinates,

    COUNT(*) FILTER (WHERE s IS NULL)
        AS missing_speed,

    COUNT(*) FILTER (WHERE a IS NULL)
        AS missing_acceleration,

    COUNT(*) FILTER (WHERE dis IS NULL)
        AS missing_displacement,

    MIN(time_gap_seconds) AS minimum_time_gap,
    MEDIAN(time_gap_seconds) AS median_time_gap,
    MAX(time_gap_seconds) AS maximum_time_gap,

    COUNT(*) FILTER (
        WHERE time_gap_seconds > 0.15
    ) AS gaps_over_015_seconds,

    MIN(s) AS minimum_speed,
    MAX(s) AS maximum_speed,
    MIN(a) AS minimum_acceleration,
    MAX(a) AS maximum_acceleration

FROM gaps

GROUP BY
    draft_year, nfl_id, drill_type,
    drill_name, attempt, event_id

ORDER BY
    nfl_id, drill_name, attempt
""").df()

print("ATTEMPT-LEVEL RECORDS:", len(attempt_quality))
print("\nOBSERVATION COUNT DISTRIBUTION")
print(attempt_quality["observations"].describe())

print("\nMISSING MOVEMENT DATA")
print(attempt_quality[[
    "missing_coordinates",
    "missing_speed",
    "missing_acceleration",
    "missing_displacement"
]].sum())

print("\nTIME GAP SUMMARY")
print(attempt_quality[[
    "median_time_gap",
    "maximum_time_gap",
    "gaps_over_015_seconds"
]].describe())

attempt_quality.to_parquet(
    AUDIT_DIR / "combine_attempt_quality.parquet",
    index=False
)

attempt_quality.to_csv(
    AUDIT_DIR / "combine_attempt_quality.csv",
    index=False
)

print("\nAttempt quality reports saved.")
```

```python

drill_quality = (
    attempt_quality
    .groupby(["drill_type", "drill_name"])
    .agg(
        players=("nfl_id", "nunique"),
        attempts=("event_id", "nunique"),
        total_observations=("observations", "sum"),
        median_observations=("observations", "median"),
        median_sampling_gap=("median_time_gap", "median"),
        maximum_sampling_gap=("maximum_time_gap", "max"),
        missing_coordinates=("missing_coordinates", "sum"),
        missing_speed=("missing_speed", "sum"),
        missing_acceleration=("missing_acceleration", "sum"),
        large_gaps=("gaps_over_015_seconds", "sum")
    )
    .reset_index()
    .sort_values(
        ["drill_type", "players"],
        ascending=[True, False]
    )
)

print(drill_quality.to_string(index=False))

drill_quality.to_csv(
    AUDIT_DIR / "combine_drill_quality_summary.csv",
    index=False
)
```

```python

attempt_features = con.execute("""
WITH wr_players AS (
    SELECT DISTINCT nfl_id
    FROM combine_results
    WHERE combine_position = 'WR'
),

ordered AS (
    SELECT
        draft_year,
        nfl_id,
        event_id,
        drill_type,
        drill_name,
        attempt,
        time,
        x, y, s, a, dis, dir,

        LAG(time) OVER w AS previous_time,
        LAG(x) OVER w AS previous_x,
        LAG(y) OVER w AS previous_y,
        LAG(dir) OVER w AS previous_dir

    FROM combine_tracking
    WHERE entity_type = 'PLAYER'
      AND nfl_id IN (SELECT nfl_id FROM wr_players)

    WINDOW w AS (
        PARTITION BY
            draft_year, nfl_id, event_id,
            drill_type, drill_name, attempt
        ORDER BY time
    )
),

movement AS (
    SELECT *,
        EPOCH(time - previous_time) AS dt,

        SQRT(
            POWER(x - previous_x, 2) +
            POWER(y - previous_y, 2)
        ) AS step_distance,

        CASE
            WHEN dir IS NOT NULL AND previous_dir IS NOT NULL
            THEN ABS(
                ((dir - previous_dir + 180) % 360 + 360) % 360
                - 180
            )
        END AS turn_angle

    FROM ordered
),

attempt_summary AS (
    SELECT
        draft_year,
        nfl_id,
        event_id,
        drill_type,
        drill_name,
        attempt,

        COUNT(*) AS observations,

        MIN(time) AS start_time,
        MAX(time) AS end_time,

        EPOCH(MAX(time) - MIN(time)) AS duration_seconds,

        AVG(s) AS mean_speed,
        MAX(s) AS peak_speed,
        QUANTILE_CONT(s, 0.95) AS p95_speed,
        STDDEV_POP(s) AS speed_std,

        AVG(a) AS mean_acceleration,
        MAX(a) AS peak_acceleration,
        MIN(a) AS minimum_acceleration,
        STDDEV_POP(a) AS acceleration_std,

        COUNT(*) FILTER (WHERE a < 0) AS negative_acceleration_rows,

        SUM(step_distance) AS coordinate_path_length,

        SUM(dis) AS summed_recorded_displacement,

        MAX(x) - MIN(x) AS x_range,
        MAX(y) - MIN(y) AS y_range,

        SUM(turn_angle) AS total_direction_change,

        MEDIAN(dt) FILTER (WHERE dt IS NOT NULL) AS median_sampling_interval,
        MAX(dt) FILTER (WHERE dt IS NOT NULL) AS maximum_sampling_interval,

        COUNT(*) FILTER (WHERE dt > 0.15) AS gaps_over_015_seconds,

        MIN(x) AS start_x,
        MIN(y) AS start_y

    FROM movement

    GROUP BY
        draft_year, nfl_id, event_id,
        drill_type, drill_name, attempt
)

SELECT
    *,

    SQRT(
        POWER(x_range, 2) +
        POWER(y_range, 2)
    ) AS bounding_box_diagonal,

    CASE
        WHEN coordinate_path_length > 0
        THEN NULL
        ELSE NULL
    END AS reserved_path_efficiency

FROM attempt_summary
ORDER BY nfl_id, drill_name, attempt
""").df()

print("ATTEMPT FEATURE SHAPE:", attempt_features.shape)
print("\nFEATURE PREVIEW")
print(attempt_features.head(10).to_string(index=False))

print("\nFEATURE MISSINGNESS")
print(
    attempt_features.isna().sum()
    .sort_values(ascending=False)
    .head(15)
)

attempt_features.to_parquet(
    FEATURE_DIR / "wr_combine_attempt_features.parquet",
    index=False
)

attempt_features.to_csv(
    FEATURE_DIR / "wr_combine_attempt_features.csv",
    index=False
)

print("\nATTEMPT FEATURES SAVED TO GOOGLE DRIVE")
```

```python

import duckdb
import numpy as np
import pandas as pd

# Corrected attempt-level feature engineering
query = """
WITH wr_players AS (
    SELECT DISTINCT nfl_id
    FROM combine_results
    WHERE UPPER(TRIM(combine_position)) = 'WR'
),

base AS (
    SELECT
        ct.draft_year,
        ct.nfl_id,
        ct.event_id,
        ct.drill_type,
        ct.drill_name,
        ct.attempt,
        ct.time,
        ct.x,
        ct.y,
        ct.s,
        ct.a,
        ct.dis,
        ct.dir
    FROM combine_tracking ct
    INNER JOIN wr_players wp
        ON ct.nfl_id = wp.nfl_id
    WHERE UPPER(TRIM(ct.entity_type)) = 'PLAYER'
),

ordered AS (
    SELECT
        *,
        LAG(time) OVER w AS prev_time,
        LAG(x) OVER w AS prev_x,
        LAG(y) OVER w AS prev_y,
        LAG(dir) OVER w AS prev_dir
    FROM base
    WINDOW w AS (
        PARTITION BY
            draft_year, nfl_id, event_id,
            drill_type, drill_name, attempt
        ORDER BY time
    )
),

steps AS (
    SELECT
        *,
        EPOCH(time - prev_time) AS dt,

        CASE
            WHEN prev_x IS NOT NULL AND prev_y IS NOT NULL
            THEN SQRT(
                POWER(x - prev_x, 2) +
                POWER(y - prev_y, 2)
            )
        END AS step_distance,

        CASE
            WHEN dir IS NOT NULL AND prev_dir IS NOT NULL
            THEN ABS(
                ((dir - prev_dir + 540) % 360) - 180
            )
        END AS angular_change
    FROM ordered
),

aggregated AS (
    SELECT
        draft_year,
        nfl_id,
        event_id,
        drill_type,
        drill_name,
        attempt,

        COUNT(*) AS observations,

        FIRST(time ORDER BY time) AS start_time,
        LAST(time ORDER BY time) AS end_time,

        EPOCH(
            LAST(time ORDER BY time) -
            FIRST(time ORDER BY time)
        ) AS duration_seconds,

        FIRST(x ORDER BY time) AS start_x,
        FIRST(y ORDER BY time) AS start_y,
        LAST(x ORDER BY time) AS end_x,
        LAST(y ORDER BY time) AS end_y,

        AVG(s) AS mean_speed,
        MAX(s) AS peak_speed,
        QUANTILE_CONT(s, 0.95) AS p95_speed,
        STDDEV_POP(s) AS speed_std,

        AVG(a) AS mean_acceleration,
        MAX(a) AS peak_acceleration,
        MIN(a) AS minimum_acceleration,
        STDDEV_POP(a) AS acceleration_std,

        SUM(step_distance) AS coordinate_path_length,
        SUM(dis) AS summed_recorded_displacement,

        SUM(angular_change) AS total_direction_change,

        MEDIAN(dt) AS median_sampling_interval,
        MAX(dt) AS maximum_sampling_interval,

        SUM(CASE WHEN dt > 0.15 THEN 1 ELSE 0 END)
            AS large_sampling_gaps

    FROM steps
    GROUP BY
        draft_year, nfl_id, event_id,
        drill_type, drill_name, attempt
),

final AS (
    SELECT
        *,
        SQRT(
            POWER(end_x - start_x, 2) +
            POWER(end_y - start_y, 2)
        ) AS net_displacement
    FROM aggregated
)

SELECT
    *,
    CASE
        WHEN coordinate_path_length > 0
        THEN net_displacement / coordinate_path_length
        ELSE NULL
    END AS path_efficiency

FROM final
"""

corrected_features = con.execute(query).df()

# Save as a new version; preserve original files
corrected_parquet = (
    FEATURE_DIR / "wr_combine_attempt_features_v2.parquet"
)

corrected_csv = (
    FEATURE_DIR / "wr_combine_attempt_features_v2.csv"
)

corrected_features.to_parquet(
    corrected_parquet, index=False
)

corrected_features.to_csv(
    corrected_csv, index=False
)

print("CORRECTED ATTEMPT FEATURES")
print("Rows:", len(corrected_features))
print("Unique players:", corrected_features.nfl_id.nunique())
print("Unique events:", corrected_features.event_id.nunique())
print("Unique drill names:", corrected_features.drill_name.nunique())

print("\nPATH EFFICIENCY SUMMARY")
display(corrected_features["path_efficiency"].describe())

print("\nNET DISPLACEMENT SUMMARY")
display(corrected_features["net_displacement"].describe())

print("\nDIRECTION CHANGE SUMMARY")
display(corrected_features["total_direction_change"].describe())

print("\nSAVED:")
print(corrected_parquet)
print(corrected_csv)
```

```python
import numpy as np
import pandas as pd
from pathlib import Path

# ---------------------------------------------------------
# 1. LOAD RAW WIDE RECEIVER COMBINE TRACKING
# ---------------------------------------------------------

wr_ids = con.execute("""
    SELECT DISTINCT nfl_id
    FROM combine_results
    WHERE UPPER(TRIM(combine_position)) = 'WR'
""").df()["nfl_id"].tolist()

raw = con.execute("""
    SELECT
        draft_year,
        nfl_id,
        event_id,
        drill_type,
        drill_name,
        attempt,
        time,
        x,
        y,
        s,
        a,
        dir
    FROM combine_tracking
    WHERE entity_type = 'PLAYER'
      AND nfl_id IN (
          SELECT DISTINCT nfl_id
          FROM combine_results
          WHERE UPPER(TRIM(combine_position)) = 'WR'
      )
    ORDER BY nfl_id, event_id, time
""").df()

print("RAW TRACKING ROWS:", len(raw))
print("UNIQUE PLAYERS:", raw["nfl_id"].nunique())

# ---------------------------------------------------------
# 2. CALCULATE COORDINATE-BASED MOVEMENT HEADING
# ---------------------------------------------------------

raw = raw.sort_values(
    ["draft_year", "nfl_id", "event_id", "attempt", "time"]
).copy()

group_cols = [
    "draft_year", "nfl_id", "event_id",
    "drill_type", "drill_name", "attempt"
]

raw["dx"] = raw.groupby(group_cols)["x"].diff()
raw["dy"] = raw.groupby(group_cols)["y"].diff()

raw["step_distance"] = np.sqrt(
    raw["dx"]**2 + raw["dy"]**2
)

# Coordinate-derived heading, in degrees
raw["coordinate_heading"] = (
    np.degrees(np.arctan2(raw["dy"], raw["dx"])) + 360
) % 360

# Circular angular difference, constrained to 0–180 degrees
def angular_difference(a, b):
    return np.abs((a - b + 180) % 360 - 180)

# Ignore extremely small coordinate steps when comparing headings
valid = raw[
    raw["dir"].notna()
    & raw["coordinate_heading"].notna()
    & (raw["step_distance"] > 0.01)
].copy()

valid["heading_error_direct"] = angular_difference(
    valid["dir"], valid["coordinate_heading"]
)

# Also test the opposite-angle convention
valid["heading_error_opposite"] = angular_difference(
    valid["dir"], (valid["coordinate_heading"] + 180) % 360
)

# ---------------------------------------------------------
# 3. COMPARE DIRECTION CONVENTIONS
# ---------------------------------------------------------

print("\nHEADING AGREEMENT")
print(
    valid[
        ["heading_error_direct", "heading_error_opposite"]
    ].describe(percentiles=[0.1, 0.25, 0.5, 0.75, 0.9]).round(2)
)

print("\nHEADING AGREEMENT BY DRILL")
heading_by_drill = (
    valid.groupby("drill_name")
    .agg(
        observations=("heading_error_direct", "size"),
        median_direct_error=("heading_error_direct", "median"),
        median_opposite_error=("heading_error_opposite", "median")
    )
    .sort_values("observations", ascending=False)
    .round(2)
)

display(heading_by_drill)

# ---------------------------------------------------------
# 4. EXAMINE FRAME-TO-FRAME DIRECTION CHANGES
# ---------------------------------------------------------

raw["direction_step_change"] = (
    raw.groupby(group_cols)["dir"]
    .transform(lambda x: x.diff())
)

raw["direction_step_change"] = (
    (raw["direction_step_change"] + 180) % 360 - 180
).abs()

direction_steps = (
    raw.groupby("drill_name")
    .agg(
        observations=("direction_step_change", "count"),
        median_step_change=("direction_step_change", "median"),
        p90_step_change=("direction_step_change",
                         lambda x: x.quantile(0.90)),
        p95_step_change=("direction_step_change",
                         lambda x: x.quantile(0.95)),
        maximum_step_change=("direction_step_change", "max")
    )
    .round(2)
)

print("\nFRAME-TO-FRAME DIRECTION CHANGE BY DRILL")
display(direction_steps)

# ---------------------------------------------------------
# 5. SAVE AUDIT OUTPUTS
# ---------------------------------------------------------

REPORT_DIR.mkdir(parents=True, exist_ok=True)

heading_by_drill.to_csv(
    REPORT_DIR / "direction_heading_agreement_by_drill.csv"
)

direction_steps.to_csv(
    REPORT_DIR / "direction_step_change_by_drill.csv"
)

print("\nSAVED DIRECTION AUDITS:")
print(REPORT_DIR / "direction_heading_agreement_by_drill.csv")
print(REPORT_DIR / "direction_step_change_by_drill.csv")
```

```python
# =========================================================
# WR-MOVEMENTIQ: COORDINATE-ANGLE CONVENTION AUDIT
# =========================================================

import numpy as np
import pandas as pd

audit = raw.copy()

# Recalculate consecutive coordinate changes
audit = audit.sort_values(
    ["draft_year", "nfl_id", "event_id", "attempt", "time"]
).copy()

groups = [
    "draft_year", "nfl_id", "event_id", "attempt"
]

audit["dx"] = audit.groupby(groups)["x"].diff()
audit["dy"] = audit.groupby(groups)["y"].diff()

audit["step_distance"] = np.sqrt(
    audit["dx"]**2 + audit["dy"]**2
)

# Keep meaningful coordinate movements
audit = audit[
    audit["dir"].notna()
    & (audit["step_distance"] > 0.01)
].copy()

# Test four common coordinate-angle conventions
conventions = {
    "atan2_dy_dx": np.degrees(
        np.arctan2(audit["dy"], audit["dx"])
    ) % 360,

    "atan2_dx_dy": np.degrees(
        np.arctan2(audit["dx"], audit["dy"])
    ) % 360,

    "atan2_minus_dy_dx": np.degrees(
        np.arctan2(-audit["dy"], audit["dx"])
    ) % 360,

    "atan2_dx_minus_dy": np.degrees(
        np.arctan2(audit["dx"], -audit["dy"])
    ) % 360
}

def circular_error(a, b):
    return np.abs((a - b + 180) % 360 - 180)

results = []

for name, heading in conventions.items():
    error = circular_error(audit["dir"], heading)

    results.append({
        "convention": name,
        "observations": len(error),
        "median_error": error.median(),
        "mean_error": error.mean(),
        "p25_error": error.quantile(0.25),
        "p75_error": error.quantile(0.75),
        "within_15_degrees_pct": (error <= 15).mean() * 100,
        "within_30_degrees_pct": (error <= 30).mean() * 100
    })

convention_results = pd.DataFrame(results).sort_values(
    "median_error"
)

print("COORDINATE ANGLE CONVENTION COMPARISON")
display(convention_results.round(2))

# ---------------------------------------------------------
# CHECK RECORDED SPEED AGAINST COORDINATE STEP SPEED
# ---------------------------------------------------------

audit["time_seconds"] = (
    audit.groupby(groups)["time"]
    .diff()
    .dt.total_seconds()
)

audit["coordinate_speed"] = (
    audit["step_distance"] / audit["time_seconds"]
)

speed_check = audit[
    (audit["time_seconds"] > 0)
    & audit["s"].notna()
    & audit["coordinate_speed"].notna()
].copy()

print("\nRECORDED SPEED VS COORDINATE-DERIVED SPEED")

print(
    speed_check[
        ["s", "coordinate_speed"]
    ].describe().round(3)
)

print(
    "\nPearson correlation:",
    round(
        speed_check["s"].corr(
            speed_check["coordinate_speed"]
        ), 4
    )
)

print(
    "Spearman correlation:",
    round(
        speed_check["s"].corr(
            speed_check["coordinate_speed"],
            method="spearman"
        ), 4
    )
)

# Save results
convention_results.to_csv(
    REPORT_DIR / "coordinate_angle_convention_audit.csv",
    index=False
)

print("\nSAVED:")
print(REPORT_DIR / "coordinate_angle_convention_audit.csv")
```

```python

# ============================================================
# WR-MOVEMENTIQ | TURNING MECHANICS VALIDATION
# ============================================================

import numpy as np
import pandas as pd

# Use the existing raw DataFrame; do not reload or overwrite it.
turn = raw.copy()

groups = [
    "draft_year", "nfl_id", "event_id",
    "drill_type", "drill_name", "attempt"
]

turn = turn.sort_values(groups + ["time"]).copy()

# Consecutive coordinate movements
turn["dx"] = turn.groupby(groups)["x"].diff()
turn["dy"] = turn.groupby(groups)["y"].diff()

turn["step_distance"] = np.sqrt(
    turn["dx"]**2 + turn["dy"]**2
)

turn["dt"] = (
    turn.groupby(groups)["time"]
    .diff()
    .dt.total_seconds()
)

# Empirically validated heading convention:
# clockwise from positive Y-axis
turn["coordinate_heading"] = (
    np.degrees(np.arctan2(turn["dx"], turn["dy"])) % 360
)

def circular_change(series):
    delta = series.diff()
    return ((delta + 180) % 360 - 180).abs()

# Compare consecutive heading changes
turn["recorded_turn"] = (
    turn.groupby(groups)["dir"]
    .transform(circular_change)
)

turn["coordinate_turn"] = (
    turn.groupby(groups)["coordinate_heading"]
    .transform(circular_change)
)

# Exclude very small coordinate movements from coordinate-turn
# interpretation because heading is unstable near zero displacement.
turn.loc[
    turn["step_distance"] <= 0.01,
    "coordinate_turn"
] = np.nan

# Attempt-level turning summary
turn_features = (
    turn.groupby(groups)
    .agg(
        observations=("dir", "count"),
        duration_seconds=("time", lambda x:
            (x.max() - x.min()).total_seconds()),

        recorded_total_turn=("recorded_turn", "sum"),
        coordinate_total_turn=("coordinate_turn", "sum"),

        recorded_median_turn=("recorded_turn", "median"),
        coordinate_median_turn=("coordinate_turn", "median"),

        recorded_p95_turn=("recorded_turn",
                           lambda x: x.quantile(0.95)),
        coordinate_p95_turn=("coordinate_turn",
                             lambda x: x.quantile(0.95)),

        recorded_max_turn=("recorded_turn", "max"),
        coordinate_max_turn=("coordinate_turn", "max"),

        coordinate_turn_observations=("coordinate_turn", "count")
    )
    .reset_index()
)

# Difference between two independent turning estimates
turn_features["total_turn_difference"] = (
    turn_features["recorded_total_turn"]
    - turn_features["coordinate_total_turn"]
).abs()

turn_features["turn_correlation"] = np.nan

# Summary by drill
drill_turn_summary = (
    turn_features.groupby("drill_name")
    .agg(
        attempts=("nfl_id", "size"),
        median_recorded_total=("recorded_total_turn", "median"),
        median_coordinate_total=("coordinate_total_turn", "median"),
        median_total_difference=("total_turn_difference", "median"),
        median_coordinate_p95=("coordinate_p95_turn", "median"),
        median_coordinate_max=("coordinate_max_turn", "median")
    )
    .sort_values("attempts", ascending=False)
    .round(2)
)

print("ATTEMPT-LEVEL TURNING FEATURES")
print("Rows:", len(turn_features))
print("Unique players:", turn_features["nfl_id"].nunique())

print("\nOVERALL TURNING COMPARISON")
display(
    turn_features[
        [
            "recorded_total_turn",
            "coordinate_total_turn",
            "total_turn_difference",
            "coordinate_p95_turn",
            "coordinate_max_turn"
        ]
    ].describe().round(2)
)

print("\nTURNING SUMMARY BY DRILL")
display(drill_turn_summary)

print("\nCORRELATION BETWEEN TOTAL TURNING MEASURES")
print(
    turn_features[
        ["recorded_total_turn", "coordinate_total_turn"]
    ].corr().round(4)
)

# Preserve audit outputs separately
turn_features.to_parquet(
    FEATURE_DIR / "wr_combine_turning_audit.parquet",
    index=False
)

turn_features.to_csv(
    FEATURE_DIR / "wr_combine_turning_audit.csv",
    index=False
)

drill_turn_summary.to_csv(
    REPORT_DIR / "turning_feature_validation_by_drill.csv"
)

print("\nTURNING AUDIT SAVED.")
```

```python

# ============================================================
# WR-MOVEMENTIQ | TURNING SENSITIVITY TO MOVEMENT SPEED
# ============================================================

import numpy as np
import pandas as pd

turn_sensitivity = raw.copy()

groups = [
    "draft_year", "nfl_id", "event_id",
    "drill_type", "drill_name", "attempt"
]

turn_sensitivity = turn_sensitivity.sort_values(
    groups + ["time"]
).copy()

# Coordinate differences
turn_sensitivity["dx"] = (
    turn_sensitivity.groupby(groups)["x"].diff()
)
turn_sensitivity["dy"] = (
    turn_sensitivity.groupby(groups)["y"].diff()
)

turn_sensitivity["step_distance"] = np.sqrt(
    turn_sensitivity["dx"]**2 +
    turn_sensitivity["dy"]**2
)

turn_sensitivity["dt"] = (
    turn_sensitivity.groupby(groups)["time"]
    .diff()
    .dt.total_seconds()
)

# Coordinate-derived speed
turn_sensitivity["coordinate_speed"] = (
    turn_sensitivity["step_distance"] /
    turn_sensitivity["dt"]
)

# Validated heading convention
turn_sensitivity["coordinate_heading"] = (
    np.degrees(
        np.arctan2(
            turn_sensitivity["dx"],
            turn_sensitivity["dy"]
        )
    ) % 360
)

# Previous observation within each attempt
turn_sensitivity["previous_heading"] = (
    turn_sensitivity.groupby(groups)["coordinate_heading"].shift(1)
)

turn_sensitivity["previous_dir"] = (
    turn_sensitivity.groupby(groups)["dir"].shift(1)
)

turn_sensitivity["previous_speed"] = (
    turn_sensitivity.groupby(groups)["s"].shift(1)
)

turn_sensitivity["previous_time"] = (
    turn_sensitivity.groupby(groups)["time"].shift(1)
)

turn_sensitivity["pair_dt"] = (
    turn_sensitivity["time"] -
    turn_sensitivity["previous_time"]
).dt.total_seconds()

# Ensure consecutive observations are approximately 0.1 sec apart
consecutive = (
    turn_sensitivity["pair_dt"].between(0.05, 0.15)
)

def angular_difference(a, b):
    return np.abs((a - b + 180) % 360 - 180)

turn_sensitivity["coordinate_turn_pair"] = angular_difference(
    turn_sensitivity["coordinate_heading"],
    turn_sensitivity["previous_heading"]
)

turn_sensitivity["recorded_turn_pair"] = angular_difference(
    turn_sensitivity["dir"],
    turn_sensitivity["previous_dir"]
)

# Speed thresholds in the recorded speed's raw units
thresholds = [0, 0.5, 1, 2, 3]

sensitivity_results = []

for threshold in thresholds:

    valid_pairs = (
        consecutive
        & turn_sensitivity["s"].notna()
        & turn_sensitivity["previous_speed"].notna()
        & turn_sensitivity["coordinate_turn_pair"].notna()
        & turn_sensitivity["recorded_turn_pair"].notna()
        & (turn_sensitivity["s"] >= threshold)
        & (turn_sensitivity["previous_speed"] >= threshold)
    )

    selected = turn_sensitivity.loc[valid_pairs].copy()

    attempt_summary = (
        selected.groupby(groups)
        .agg(
            recorded_total=("recorded_turn_pair", "sum"),
            coordinate_total=("coordinate_turn_pair", "sum"),
            coordinate_p95=("coordinate_turn_pair",
                            lambda x: x.quantile(0.95)),
            valid_pairs=("coordinate_turn_pair", "size")
        )
        .reset_index()
    )

    correlation = attempt_summary[
        ["recorded_total", "coordinate_total"]
    ].corr().iloc[0, 1]

    sensitivity_results.append({
        "speed_threshold": threshold,
        "valid_observations": len(selected),
        "attempts": len(attempt_summary),
        "median_recorded_turn": attempt_summary["recorded_total"].median(),
        "median_coordinate_turn": attempt_summary["coordinate_total"].median(),
        "median_absolute_difference": (
            attempt_summary["recorded_total"] -
            attempt_summary["coordinate_total"]
        ).abs().median(),
        "median_coordinate_p95": attempt_summary["coordinate_p95"].median(),
        "turn_correlation": correlation
    })

sensitivity_table = pd.DataFrame(sensitivity_results)

print("TURNING SENSITIVITY BY SPEED THRESHOLD")
display(sensitivity_table.round(3))

# ------------------------------------------------------------
# CHECK HEADING AGREEMENT ACROSS SPEED BINS
# ------------------------------------------------------------

heading_error = angular_difference(
    turn_sensitivity["dir"],
    turn_sensitivity["coordinate_heading"]
)

turn_sensitivity["heading_error"] = heading_error

turn_sensitivity["speed_band"] = pd.cut(
    turn_sensitivity["s"],
    bins=[-0.001, 0.5, 1, 2, 3, 20],
    labels=["0–0.5", "0.5–1", "1–2", "2–3", "Above 3"]
)

speed_heading_audit = (
    turn_sensitivity.dropna(subset=["heading_error"])
    .groupby("speed_band", observed=True)
    .agg(
        observations=("heading_error", "size"),
        median_heading_error=("heading_error", "median"),
        p90_heading_error=("heading_error",
                           lambda x: x.quantile(0.90))
    )
    .reset_index()
)

print("\nHEADING ERROR BY RECORDED SPEED")
display(speed_heading_audit.round(3))

# Save audit outputs
sensitivity_table.to_csv(
    REPORT_DIR / "turning_speed_threshold_sensitivity.csv",
    index=False
)

speed_heading_audit.to_csv(
    REPORT_DIR / "heading_error_by_speed.csv",
    index=False
)

print("\nSAVED:")
print(REPORT_DIR / "turning_speed_threshold_sensitivity.csv")
print(REPORT_DIR / "heading_error_by_speed.csv")
```

```python

# ============================================================
# WR-MOVEMENTIQ | FINAL CANDIDATE TURNING FEATURES
# ============================================================

import numpy as np
import pandas as pd

# Use previously constructed turn_sensitivity DataFrame
df = turn_sensitivity.copy()

groups = [
    "draft_year", "nfl_id", "event_id",
    "drill_type", "drill_name", "attempt"
]

# Define comparable consecutive observations
valid_pairs = (
    df["pair_dt"].between(0.05, 0.15)
    & df["s"].notna()
    & df["previous_speed"].notna()
    & df["coordinate_turn_pair"].notna()
    & df["recorded_turn_pair"].notna()
)

df["valid_turn_pair"] = valid_pairs

# Build attempt-level features for two thresholds
candidate_features = []

for threshold in [1.0, 2.0]:

    selected = df[
        valid_pairs
        & (df["s"] >= threshold)
        & (df["previous_speed"] >= threshold)
    ].copy()

    features = (
        selected.groupby(groups)
        .agg(
            valid_turn_pairs=("coordinate_turn_pair", "size"),

            coordinate_total_turn=(
                "coordinate_turn_pair", "sum"
            ),

            coordinate_mean_turn=(
                "coordinate_turn_pair", "mean"
            ),

            coordinate_median_turn=(
                "coordinate_turn_pair", "median"
            ),

            coordinate_p95_turn=(
                "coordinate_turn_pair",
                lambda x: x.quantile(0.95)
            ),

            coordinate_max_turn=(
                "coordinate_turn_pair", "max"
            ),

            recorded_total_turn=(
                "recorded_turn_pair", "sum"
            ),

            recorded_mean_turn=(
                "recorded_turn_pair", "mean"
            )
        )
        .reset_index()
    )

    features["speed_threshold"] = threshold

    candidate_features.append(features)

turning_candidates = pd.concat(
    candidate_features,
    ignore_index=True
)

# Add attempt duration and valid observation coverage
attempt_info = (
    df.groupby(groups)
    .agg(
        attempt_duration_seconds=(
            "time",
            lambda x: (x.max() - x.min()).total_seconds()
        ),
        total_observations=("time", "size")
    )
    .reset_index()
)

turning_candidates = turning_candidates.merge(
    attempt_info,
    on=groups,
    how="left"
)

turning_candidates["turning_rate_degrees_per_second"] = (
    turning_candidates["coordinate_total_turn"]
    / turning_candidates["attempt_duration_seconds"]
)

turning_candidates["valid_pair_coverage"] = (
    turning_candidates["valid_turn_pairs"]
    / (turning_candidates["total_observations"] - 1)
)

# Check structural integrity
print("TURNING CANDIDATE DATASET")
print("Rows:", len(turning_candidates))
print("Players:", turning_candidates["nfl_id"].nunique())
print("Thresholds:", sorted(
    turning_candidates["speed_threshold"].unique()
))

print("\nCOVERAGE BY SPEED THRESHOLD")
display(
    turning_candidates.groupby("speed_threshold")
    .agg(
        attempts=("nfl_id", "size"),
        median_valid_pairs=("valid_turn_pairs", "median"),
        median_coverage=("valid_pair_coverage", "median"),
        median_turning_rate=(
            "turning_rate_degrees_per_second", "median"
        )
    ).round(3)
)

print("\nFEATURE SUMMARY BY THRESHOLD")
display(
    turning_candidates.groupby("speed_threshold")[
        [
            "coordinate_total_turn",
            "coordinate_mean_turn",
            "coordinate_p95_turn",
            "turning_rate_degrees_per_second"
        ]
    ].median().round(3)
)

# Save separately
turning_candidates.to_parquet(
    FEATURE_DIR / "wr_combine_turning_candidate_features.parquet",
    index=False
)

turning_candidates.to_csv(
    FEATURE_DIR / "wr_combine_turning_candidate_features.csv",
    index=False
)

print("\nSAVED:")
print(
    FEATURE_DIR /
    "wr_combine_turning_candidate_features.parquet"
)
print(
    FEATURE_DIR /
    "wr_combine_turning_candidate_features.csv"
)
```

```python
import pandas as pd
import numpy as np
from pathlib import Path

PROJECT_ROOT = Path(
    "/content/drive/MyDrive/NFL-BigDataBowl-2027"
)

FEATURE_PATH = (
    PROJECT_ROOT
    / "data/features/wr_combine_turning_candidate_features.parquet"
)

AUDIT_DIR = PROJECT_ROOT / "reports/data_audit"
AUDIT_DIR.mkdir(parents=True, exist_ok=True)

turning = pd.read_parquet(FEATURE_PATH)

# --------------------------------------------------
# 1. COVERAGE DISTRIBUTION BY DRILL AND THRESHOLD
# --------------------------------------------------

coverage_audit = (
    turning
    .groupby(
        ["speed_threshold", "drill_name"],
        dropna=False
    )
    .agg(
        attempts=("nfl_id", "size"),
        players=("nfl_id", "nunique"),
        coverage_min=("valid_pair_coverage", "min"),
        coverage_p10=("valid_pair_coverage",
                      lambda x: x.quantile(0.10)),
        coverage_median=("valid_pair_coverage", "median"),
        coverage_p90=("valid_pair_coverage",
                      lambda x: x.quantile(0.90)),
        below_70_pct=("valid_pair_coverage",
                      lambda x: (x < 0.70).mean() * 100),
        below_80_pct=("valid_pair_coverage",
                      lambda x: (x < 0.80).mean() * 100)
    )
    .reset_index()
)

coverage_audit.to_csv(
    AUDIT_DIR / "turning_coverage_distribution_by_drill.csv",
    index=False
)

print("COVERAGE DISTRIBUTION BY DRILL")
display(
    coverage_audit.sort_values(
        ["speed_threshold", "coverage_median"]
    ).round(3)
)

# --------------------------------------------------
# 2. IDENTIFY LOW-COVERAGE ATTEMPTS
# --------------------------------------------------

attempt_keys = [
    c for c in [
        "draft_year",
        "nfl_id",
        "drill_type",
        "drill_name",
        "attempt",
        "event_id"
    ]
    if c in turning.columns
]

primary = turning[
    turning["speed_threshold"] == 1.0
].copy()

low_coverage = primary[
    primary["valid_pair_coverage"] < 0.70
].copy()

low_coverage = low_coverage.sort_values(
    ["drill_name", "valid_pair_coverage"]
)

low_coverage.to_csv(
    AUDIT_DIR / "turning_low_coverage_attempts_threshold_1.csv",
    index=False
)

print("\nLOW-COVERAGE ATTEMPTS AT PRIMARY THRESHOLD")
print("Attempts below 70%:", len(low_coverage))
print("Percentage:",
      round(len(low_coverage) / len(primary) * 100, 2), "%")

display(
    low_coverage[
        attempt_keys + [
            "valid_turn_pairs",
            "total_observations",
            "valid_pair_coverage",
            "coordinate_total_turn",
            "coordinate_p95_turn"
        ]
    ].head(30).round(3)
)

# --------------------------------------------------
# 3. LOW-COVERAGE COUNTS BY DRILL
# --------------------------------------------------

low_coverage_by_drill = (
    primary
    .assign(
        below_70=lambda x: x["valid_pair_coverage"] < 0.70,
        below_80=lambda x: x["valid_pair_coverage"] < 0.80
    )
    .groupby("drill_name")
    .agg(
        attempts=("nfl_id", "size"),
        below_70=("below_70", "sum"),
        below_80=("below_80", "sum")
    )
    .reset_index()
)

low_coverage_by_drill["below_70_pct"] = (
    100 * low_coverage_by_drill["below_70"]
    / low_coverage_by_drill["attempts"]
)

low_coverage_by_drill["below_80_pct"] = (
    100 * low_coverage_by_drill["below_80"]
    / low_coverage_by_drill["attempts"]
)

low_coverage_by_drill.to_csv(
    AUDIT_DIR / "turning_low_coverage_summary.csv",
    index=False
)

print("\nLOW-COVERAGE SUMMARY")
display(
    low_coverage_by_drill.sort_values(
        "below_70_pct", ascending=False
    ).round(2)
)

print("\nSAVED REPORTS:")
print(AUDIT_DIR / "turning_coverage_distribution_by_drill.csv")
print(AUDIT_DIR / "turning_low_coverage_attempts_threshold_1.csv")
print(AUDIT_DIR / "turning_low_coverage_summary.csv")
```

```python
import pandas as pd
import numpy as np
from pathlib import Path

ROOT = Path("/content/drive/MyDrive/NFL-BigDataBowl-2027")

FEATURE_DIR = ROOT / "data/features"
AUDIT_DIR = ROOT / "reports/data_audit"

combine = pd.read_parquet(
    FEATURE_DIR / "wr_combine_attempt_features_v2.parquet"
)

turning_all = pd.read_parquet(
    FEATURE_DIR / "wr_combine_turning_candidate_features.parquet"
)

game = pd.read_parquet(
    FEATURE_DIR / "wr_game_movement_features.parquet"
)

# --------------------------------------------------
# 1. SELECT PRIMARY TURNING THRESHOLD
# --------------------------------------------------

turning = turning_all.loc[
    turning_all["speed_threshold"].eq(1.0)
].copy()

attempt_key = [
    "draft_year",
    "nfl_id",
    "drill_type",
    "drill_name",
    "attempt",
    "event_id"
]

assert not combine.duplicated(attempt_key).any()
assert not turning.duplicated(attempt_key).any()

print("COMBINE ATTEMPTS:", len(combine))
print("PRIMARY TURNING ATTEMPTS:", len(turning))

# --------------------------------------------------
# 2. CHECK KEY MATCHING BEFORE MERGING
# --------------------------------------------------

combine_keys = set(
    map(tuple, combine[attempt_key].to_numpy())
)

turning_keys = set(
    map(tuple, turning[attempt_key].to_numpy())
)

print("Keys only in Combine:", len(combine_keys - turning_keys))
print("Keys only in Turning:", len(turning_keys - combine_keys))
print("Keys shared:", len(combine_keys & turning_keys))

if combine_keys != turning_keys:
    raise ValueError("Combine and turning attempt keys do not match.")

# --------------------------------------------------
# 3. MERGE
# --------------------------------------------------

turning_columns = attempt_key + [
    "valid_turn_pairs",
    "coordinate_total_turn",
    "coordinate_mean_turn",
    "coordinate_median_turn",
    "coordinate_p95_turn",
    "coordinate_max_turn",
    "recorded_total_turn",
    "recorded_mean_turn",
    "speed_threshold",
    "turning_rate_degrees_per_second",
    "valid_pair_coverage"
]

combine_enriched = combine.merge(
    turning[turning_columns],
    on=attempt_key,
    how="left",
    validate="one_to_one"
)

assert len(combine_enriched) == len(combine)
assert combine_enriched["coordinate_total_turn"].notna().all()

# --------------------------------------------------
# 4. QUALITY FLAGS
# --------------------------------------------------

combine_enriched["turning_coverage_flag"] = np.select(
    [
        combine_enriched["valid_pair_coverage"] < 0.70,
        combine_enriched["valid_pair_coverage"] < 0.80
    ],
    [
        "LOW_BELOW_70",
        "MODERATE_70_TO_80"
    ],
    default="GOOD_80_OR_ABOVE"
)

print("\nENRICHED COMBINE DATASET")
print("Rows:", len(combine_enriched))
print("Players:", combine_enriched["nfl_id"].nunique())
print("Unique attempts:",
      combine_enriched[attempt_key].drop_duplicates().shape[0])

print("\nCOVERAGE FLAGS")
display(
    combine_enriched["turning_coverage_flag"]
    .value_counts()
    .rename_axis("coverage_flag")
    .reset_index(name="attempts")
)

# --------------------------------------------------
# 5. SAVE NEW DATASET
# --------------------------------------------------

parquet_path = (
    FEATURE_DIR / "wr_combine_attempt_features_enriched.parquet"
)

csv_path = (
    FEATURE_DIR / "wr_combine_attempt_features_enriched.csv"
)

combine_enriched.to_parquet(parquet_path, index=False)
combine_enriched.to_csv(csv_path, index=False)

print("\nSAVED:")
print(parquet_path)
print(csv_path)

# --------------------------------------------------
# 6. GAME SEASON AND PLAYER COHORT AUDIT
# --------------------------------------------------

combine_players = (
    combine[["nfl_id", "draft_year"]]
    .drop_duplicates()
)

game_players = (
    game[["nfl_id"]]
    .drop_duplicates()
)

player_overlap = combine_players.merge(
    game_players,
    on="nfl_id",
    how="left",
    indicator=True,
    validate="one_to_one"
)

print("\nCOMBINE-TO-GAME PLAYER LINKAGE")
display(player_overlap["_merge"].value_counts())

# Attach draft year to game records.
game_cohort = game.merge(
    combine_players,
    on="nfl_id",
    how="inner",
    validate="many_to_one"
)

# Only retain seasons on or after the player's draft year.
game_post_draft = game_cohort.loc[
    game_cohort["season"] >= game_cohort["draft_year"]
].copy()

cohort_audit = (
    game_post_draft
    .groupby(["draft_year", "season"])
    .agg(
        players=("nfl_id", "nunique"),
        receiver_plays=("play_id", "size"),
        games=("game_id", "nunique")
    )
    .reset_index()
)

print("\nPOST-DRAFT GAME COVERAGE")
display(cohort_audit)

print("\nGAME POSITION DISTRIBUTION")
display(
    game_post_draft["position"]
    .value_counts(dropna=False)
    .rename_axis("position")
    .reset_index(name="rows")
)

cohort_audit.to_csv(
    AUDIT_DIR / "post_draft_game_coverage_by_cohort.csv",
    index=False
)

print("\nPOST-DRAFT GAME ROWS:", len(game_post_draft))
print("POST-DRAFT PLAYERS:", game_post_draft["nfl_id"].nunique())
```

```python
import pandas as pd
from pathlib import Path

ROOT = Path("/content/drive/MyDrive/NFL-BigDataBowl-2027")

FEATURE_DIR = ROOT / "data/features"
RAW_DIR = ROOT / "data/raw/nfl-big-data-bowl-2027"
AUDIT_DIR = ROOT / "reports/data_audit"

combine = pd.read_parquet(
    FEATURE_DIR / "wr_combine_attempt_features_v2.parquet"
)

game = pd.read_parquet(
    FEATURE_DIR / "wr_game_movement_features.parquet"
)

players = pd.read_csv(
    RAW_DIR / "players.csv"
)

# One Combine record per player
combine_players = (
    combine[["nfl_id", "draft_year"]]
    .drop_duplicates()
)

# Player coverage in the game movement dataset
game_player_coverage = (
    game.groupby("nfl_id")
    .agg(
        game_seasons=("season", "nunique"),
        first_game_season=("season", "min"),
        last_game_season=("season", "max"),
        receiver_plays=("play_id", "size"),
        games=("game_id", "nunique"),
        positions=("position", lambda x: ", ".join(
            sorted(x.dropna().unique())
        ))
    )
    .reset_index()
)

# Join player register
player_register = players[
    ["nfl_id", "display_name", "nfl_position",
     "college_name", "draft_round", "draft_overall_pick"]
].drop_duplicates("nfl_id")

coverage = (
    combine_players
    .merge(player_register, on="nfl_id", how="left")
    .merge(
        game_player_coverage,
        on="nfl_id",
        how="left",
        indicator=True
    )
)

# Unmatched Combine players
unmatched = coverage[
    coverage["_merge"] == "left_only"
].copy()

matched = coverage[
    coverage["_merge"] == "both"
].copy()

print("COMBINE PLAYERS:", len(coverage))
print("MATCHED PLAYERS:", len(matched))
print("UNMATCHED PLAYERS:", len(unmatched))

print("\nUNMATCHED PLAYER REGISTER")
display(
    unmatched[
        [
            "nfl_id",
            "display_name",
            "draft_year",
            "nfl_position",
            "college_name",
            "draft_round",
            "draft_overall_pick"
        ]
    ].sort_values(["draft_year", "display_name"])
)

# Save detailed player-level linkage audit
coverage.to_csv(
    AUDIT_DIR / "combine_game_player_linkage_detail.csv",
    index=False
)

unmatched.to_csv(
    AUDIT_DIR / "unmatched_combine_receivers.csv",
    index=False
)

# --------------------------------------------------
# GAME EXPOSURE DISTRIBUTION
# --------------------------------------------------

print("\nMATCHED PLAYER GAME EXPOSURE")

display(
    matched.groupby("draft_year")
    .agg(
        players=("nfl_id", "nunique"),
        median_receiver_plays=("receiver_plays", "median"),
        mean_receiver_plays=("receiver_plays", "mean"),
        min_receiver_plays=("receiver_plays", "min"),
        max_receiver_plays=("receiver_plays", "max"),
        median_games=("games", "median")
    )
    .round(2)
)

# --------------------------------------------------
# PLAYER-SEASON EXPOSURE
# --------------------------------------------------

player_season_exposure = (
    game.merge(
        combine_players,
        on="nfl_id",
        how="inner",
        validate="many_to_one"
    )
    .query("season >= draft_year")
    .groupby(["draft_year", "season"])
    .agg(
        players=("nfl_id", "nunique"),
        receiver_plays=("play_id", "size"),
        games=("game_id", "nunique"),
        median_plays_per_player=(
            "nfl_id",
            lambda x: x.value_counts().median()
        )
    )
    .reset_index()
)

print("\nPLAYER-SEASON EXPOSURE")
display(player_season_exposure)

player_season_exposure.to_csv(
    AUDIT_DIR / "player_season_exposure_audit.csv",
    index=False
)

print("\nAUDIT REPORTS SAVED.")
```
